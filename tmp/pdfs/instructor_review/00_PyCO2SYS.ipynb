### One worked example: TA and DIC as inputs

For a separate sample, prescribe TA = 2300 and DIC = 2000 µmol/kg.
`par1_type=1` means TA and `par2_type=2` means DIC. The result is a dictionary:
select the output you need by its key. Adapt this call to the input pair and
conditions required by the exercise.

```python
example = pyco2.sys(
    par1=2300.0, par1_type=1,  # TA, umol/kg
    par2=2000.0, par2_type=2,  # DIC, umol/kg
    temperature=TEMPERATURE_C,
    salinity=SALINITY,
    pressure=PRESSURE_DBAR,
    **config.chemistry,
)
print(f"Seawater-scale pH: {float(example['pH']):.3f}")
print(f"DIC: {float(example['dic']):.1f} umol/kg")
print(f"Aragonite saturation: {float(example['saturation_aragonite']):.3f}")
print(f"Calcite saturation: {float(example['saturation_calcite']):.3f}")
```

### Instructor reference

<div style="background-color: #f3eefb; color: #38224f; border-left: 4px solid #7952a8; padding: 12px 16px; margin: 16px 0; border-radius: 4px;">

**Instructor reference**

All answers below use PyCO2SYS 1.8.3.4 and the unmodified `config.chemistry` choices.

```python
def equilibrium(xco2, temperature=TEMPERATURE_C, salinity=SALINITY):
    return pyco2.sys(
        par1=INITIAL_TA, par1_type=1,
        par2=xco2, par2_type=9,
        temperature=temperature, salinity=salinity,
        pressure=PRESSURE_DBAR, **config.chemistry,
    )


def show_state(label, result):
    print(f"{label}: pH = {float(result['pH']):.3f}; "
          f"DIC = {float(result['dic']):.1f} umol/kg; "
          f"TA = {float(result['alkalinity']):.1f} umol/kg; "
          f"Omega_aragonite = {float(result['saturation_aragonite']):.3f}; "
          f"Omega_calcite = {float(result['saturation_calcite']):.3f}")
```

## b) Temperature sensitivity

<div style="background-color: #edf5ff; color: #173b61; border-left: 4px solid #3977b8; padding: 12px 16px; margin: 16px 0; border-radius: 4px;">

**Question — part b**

How would pH change if the ocean temperature were **5 °C** or **25 °C**?
Keep TA = 2100 µmol/kg, salinity = 35 and xCO₂ = 430 ppm. Compare with 15 °C.

</div>

```python
# Your calculation
```

```python
temperature_states = {}
for temperature in (5.0, 15.0, 25.0):
    state = equilibrium(430.0, temperature=temperature)
    temperature_states[temperature] = state
    print(f"T = {temperature:g} degC: pH = {float(state['pH']):.3f}; "
          f"change from 15 degC = {float(state['pH'] - present['pH']):+.3f}")
```

<!-- BEGIN SOLUTION -->
<div style="background-color: #f3eefb; color: #38224f; border-left: 4px solid #7952a8; padding: 12px 16px; margin: 16px 0; border-radius: 4px;">

**Instructor answer**

| Temperature (°C) | pH | Change from 15 °C |
| --- | ---: | ---: |
| 5 | 7.971 | −0.011 |
| 15 | 7.982 | 0.000 |
| 25 | 7.985 | +0.003 |

The direct pH effect is small **at fixed TA and atmospheric xCO₂**. Changes in
CO₂ solubility and acid–base equilibria partly offset one another while DIC
adjusts. This does not imply that warming a closed sample at fixed DIC has the
same effect, or that temperature has little effect on mineral saturation or biology.

## d) Glacial and interglacial CO₂

<div style="background-color: #edf5ff; color: #173b61; border-left: 4px solid #3977b8; padding: 12px 16px; margin: 16px 0; border-radius: 4px;">

**Question — part d**

Calculate pH for a glacial atmosphere with **180 ppm** CO₂ and an interglacial
atmosphere with **270 ppm** CO₂. Would temperature differences substantially
alter the comparison? Use 15 °C for both first, then use the 5–25 °C range
from (b) as a sensitivity check; it is not a reconstructed climate history.

</div>

```python
# Your calculation
```

```python
climate_states = {}
for xco2 in (180.0, 270.0):
    for temperature in (5.0, 15.0, 25.0):
        state = equilibrium(xco2, temperature=temperature)
        climate_states[xco2, temperature] = state
        print(f"xCO2 = {xco2:g} ppm, T = {temperature:g} degC: "
              f"pH = {float(state['pH']):.3f}")
```

<!-- BEGIN SOLUTION -->
<div style="background-color: #f3eefb; color: #38224f; border-left: 4px solid #7952a8; padding: 12px 16px; margin: 16px 0; border-radius: 4px;">

**Instructor answer**

| Prescribed CO₂ (ppm) | pH at 5 °C | pH at 15 °C | pH at 25 °C |
| --- | ---: | ---: | ---: |
| 180 (glacial) | 8.299 | 8.294 | 8.281 |
| 270 (interglacial) | 8.150 | 8.153 | 8.147 |

At 15 °C, glacial pH exceeds interglacial pH by **0.142**. Temperature sensitivity
over this deliberately broad range is much smaller than the CO₂ effect and does
not reverse it. For example, 180 ppm at 5 °C versus 270 ppm at 15 °C gives a
difference of 0.146. These conditional chemistry comparisons leave changes in
real-ocean circulation, salinity and alkalinity outside the calculation.

The endpoint is prescribed, not a new forecast or a transient model result.

## h) Alkalinity needed to restore aragonite saturation

<div style="background-color: #edf5ff; color: #173b61; border-left: 4px solid #3977b8; padding: 12px 16px; margin: 16px 0; border-radius: 4px;">

**Question — part h**

At **935 ppm** CO₂, calculate the total alkalinity required to restore the
**present-day aragonite saturation from (f)**. Report both the required TA and
its increase above 2100 µmol/kg. Use 15 °C for the main answer and also report
the 18 °C alternative from (g). Does restoring saturation restore present pH?
Explain what this calculation can and cannot establish about alkalinity addition
as a way to protect coral reefs.

</div>

```python
# Your calculation
```

```python
oae_states = {}
for temperature in (15.0, 18.0):
    state = pyco2.sys(
        par1=target_aragonite, par1_type=11,  # prescribed Omega_aragonite
        par2=935.0, par2_type=9,             # prescribed dry-air xCO2
        temperature=temperature, salinity=SALINITY,
        pressure=PRESSURE_DBAR, **config.chemistry,
    )
    oae_states[temperature] = state
    show_state(f'Restored saturation at {temperature:g} degC', state)
    print(f"  Required TA enhancement: {float(state['alkalinity']) - INITIAL_TA:.1f} umol/kg")
    print(f"  pH change from present: {float(state['pH'] - present['pH']):+.3f}")

    # Independently re-solve with inferred TA + prescribed CO2 to check the inversion.
    check = pyco2.sys(
        par1=float(state['alkalinity']), par1_type=1,
        par2=935.0, par2_type=9,
        temperature=temperature, salinity=SALINITY,
        pressure=PRESSURE_DBAR, **config.chemistry,
    )
    assert abs(float(check['saturation_aragonite']) - target_aragonite) < 1e-8
    assert abs(float(check['pH'] - state['pH'])) < 1e-8
```

<div style="background-color: #f3eefb; color: #38224f; border-left: 4px solid #7952a8; padding: 12px 16px; margin: 16px 0; border-radius: 4px;">

**Instructor answer**

| Temperature | Required TA (µmol/kg) | Added TA (µmol/kg) | pH | DIC (µmol/kg) |
| --- | ---: | ---: | ---: | ---: |
| 15 °C | 2922.4 | **822.4** | 7.813 | 2779.1 |
| 18 °C | 2737.2 | **637.2** | 7.791 | 2590.2 |

Both states reproduce the target aragonite saturation of **1.975** by construction.
For the main 15 °C answer, TA must rise by about **39%**. The warmer alternative
requires less TA to attain the same saturation state. Neither restores present
pH (7.982): saturation and pH are different properties, and restoring one does
not fix the other.

The target saturation is **prescribed**, and TA is **inferred**. The final assertions
check the chemistry inversion; they do not independently validate alkalinity
enhancement as a reef intervention. DIC rises as the water equilibrates with the
prescribed atmosphere. There is no closed ocean–atmosphere carbon inventory here,
and this is a concentration increment, not a total dose or addition rate. A dose
would require the treated water mass and a specified material; adding carbonate
mineral also adds DIC, unlike an idealized pure-TA addition.

Higher equilibrium DIC is consistent with potential CO₂ uptake, but these states
do not quantify uptake rates, net atmospheric removal, mixing, precipitation,
cost or ecological feasibility. Restored saturation also does not remove thermal
stress or guarantee coral survival. A coupled process model and site-specific
evidence would be needed to assess an intervention.

### One worked example: TA and DIC as inputs

For a separate sample, prescribe TA = 2300 and DIC = 2000 µmol/kg.
`par1_type=1` means TA and `par2_type=2` means DIC. The result is a dictionary:
select the output you need by its key. Adapt this call to the input pair and
conditions required by the exercise.

```python
example = pyco2.sys(
    par1=2300.0, par1_type=1,  # TA, umol/kg
    par2=2000.0, par2_type=2,  # DIC, umol/kg
    temperature=TEMPERATURE_C,
    salinity=SALINITY,
    pressure=PRESSURE_DBAR,
    **config.chemistry,
)
print(f"Seawater-scale pH: {float(example['pH']):.3f}")
print(f"DIC: {float(example['dic']):.1f} umol/kg")
print(f"Aragonite saturation: {float(example['saturation_aragonite']):.3f}")
print(f"Calcite saturation: {float(example['saturation_calcite']):.3f}")
```
